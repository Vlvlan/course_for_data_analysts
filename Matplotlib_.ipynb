# Wykres słupkowy - bar

```python
piwa = ["harnas", "kasztelan", "beczkowe", "lech"]
liczba_piw = [112, 80, 40, 200]
```

```python
plt.bar(piwa, liczba_piw)
plt.ylabel("Ilość sprzedanych piw")
plt.xlabel("Rodzaje piw")
plt.title("Wyrkres")
plt.savefig("test_wykres", format="pdf")
plt.show()
```

```python
a = 1
b = 5

x = range(0, 15)
y = []

for i in x:
  y.append(a * i + b)

fig, ax = plt.subplots(
    figsize=(8, 6),
    dpi=100,
    facecolor="grey",

)
ax.plot(x,y, color="#07a832", linewidth=2, linestyle="-", marker="*", markerfacecolor="blue", alpha=0.2)
plt.show()
```

# Wykres kołowo - pie

```python
import matplotlib.pyplot as plt
```

```python
owoce = ["jabkła", "gruszki", "arbuz", "mandarynki", "babana"]
owoce_liczby = [20, 24, 50, 12, 18]
lista_kolorow = ["red", "yellow", "green", "orange", "#FFD580"]
```

```python
fig, ax = plt.subplots(figsize=(10,8))
ax.pie(owoce_liczby, labels=owoce, autopct="%.2f%%", colors=lista_kolorow,
       explode=[0, 0.1, 0, 0, 0], shadow=True, radius=0.8)

ax.set_title("Sprzedaż dniowa owoców w kg", fontsize=16, fontweight="bold", color="crimson")

ax.legend(owoce, title="Podział owoców", loc="upper left")
plt.show()
```

```python
owoce = ["jabłka", "banany", "gruszki", "sliwy"]
poniedzialek = [25, 40, 15, 30]
wtorek = [30, 35, 20, 25]
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10,8))

ax1.pie(poniedzialek, labels=owoce, autopct="%.0f%%")
ax1.set_title("Sprzedaz poniedziałkowa")

ax2.pie(wtorek, labels=owoce, autopct="%.0f%%")
ax2.set_title("Sprzedaz wtorkowa")

fig.suptitle("Porównanie sprzedazy dwóćh dni", fontsize=20, color="purple", fontweight="bold")

plt.show()

```

# Wykres puntkowy - scatter

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
may_customers = np.array([125, 111, 150, 147, 135, 120, 99, 85, 90, 100, 123])
may_temp = np.array([25, 27, 32, 31, 31, 29, 27, 25, 24, 26, 28])

october_customers = np.array([90, 100, 98, 85, 70, 75, 71, 68, 80, 71, 63])
october_temp = np.array([22, 23, 22, 21, 20, 19, 17, 17, 16, 17, 15])
```

```python
plt.scatter(may_temp, may_customers)
plt.show()
```

```python
plt.scatter(october_temp, october_customers)
plt.show()
```

```python
plt.scatter(may_temp, may_customers, s=60, color="green", alpha=0.5, label="May")
plt.scatter(october_temp, october_customers, s=50, color="crimson", label="October")
plt.xlabel("Temperatura stopnie C")
plt.ylabel("Ilosc klientów")
plt.title("Ruch w lodziarni")
plt.legend()
plt.show()
```

# HISTOGRAM - hist

```python
dane = np.array([np.random.randint(100, high=150, size=30, dtype=int)])
dane = dane.reshape(-1)
```

```python
dane
```

```python
bins_ = list(range(100, 150, 5))
fig, ax = plt.subplots()
ax.hist(dane, bins=bins_, edgecolor="black", linewidth=0.5)

ax.grid(axis="y", linestyle=":", alpha=0.8, color="black")
ax.set_axisbelow(True)
plt.show()
```

# Wykres słupkowy poziomy - barh

```python
punkty = [50, 40, 70, 96, 78, 56, 49, 39]
klasy = ["klasa_a", "klasa_b", "klasa_c", "klasa_d", "klasa_e", "klasa_g", "klasa_h", "klasa_i"]
```

```python
plt.barh(klasy, punkty, height=0.6, align="center", color='y', ec="r", lw=0.5)
plt.xlabel("Średnia liczba punktów na egzaminie")
plt.grid(axis="x", linestyle="--", alpha=0.8, color="black")
plt.show()
```

```python
plt.barh(["A", "B", "C"], [5,2,7], left=[0,1,2])
plt.show()
```
