# PRAKTYKA

3 klasy napisały ten sam test na którym można było zdobyć od 0 do 100 punktów,
każda klasa miała po 20 uczniów. Napis program który porówna wyniki trzech klas, w tym celu stosuj się do poniższych instrukcji.

Stwórz tablice 3d która będzie miała 3 bloki po 20 wartośći w przedziale od 0 do 100 i następnie:
- Podziel tablice na 3 osobne obiekty
- Podaj ile elementów jest w każdym obiekcie
- Wypisz średnią dla każdego obiektu
- Wypisz największą i najmniejszą wartość dla każdego obiektu
- Podaj ile elementów z każdego obiektu miało wartość większą niż 50

```python
import numpy as np
```

```python
rng = np.random.default_rng()
```

```python
test = rng.integers(low=0, high=101, size=(3,4,5))
```

```python
print(test)
```

```python
klasa_a, klasa_b, klasa_c = test[0], test[1], test[2]
```

```python
print(klasa_a)
```

```python
print(klasa_b)
```

```python
print(klasa_c)
```

```python
tekst_a = f"""
\tKLASA A\n
Ilość uczniów : {klasa_a.size}
Średnia klasy : {np.mean(klasa_a)}
Najlepszy indywidualny wynik : {np.max(klasa_a)}
Najgorszy indywidualny wynik : {np.min(klasa_a)}
Ilość uczniów która osiągneła wynik powyżej 50% : {klasa_a[klasa_a > 50].size}"""
```

```python
print(tekst_a)
```

```python
tekst_b = f"""
\tKLASA B\n
Ilość uczniów : {klasa_b.size}
Średnia klasy : {np.mean(klasa_b)}
Najlepszy indywidualny wynik : {np.max(klasa_b)}
Najgorszy indywidualny wynik : {np.min(klasa_b)}
Ilość uczniów która osiągneła wynik powyżej 50% : {klasa_b[klasa_b > 50].size}"""
```

```python
tekst_c = f"""
\tKLASA C\n
Ilość uczniów : {klasa_c.size}
Średnia klasy : {np.mean(klasa_c)}
Najlepszy indywidualny wynik : {np.max(klasa_c)}
Najgorszy indywidualny wynik : {np.min(klasa_c)}
Ilość uczniów która osiągneła wynik powyżej 50% : {klasa_c[klasa_c > 50].size}"""
```

```python
print(tekst_a, tekst_b, tekst_c, sep="\n\n")
```
