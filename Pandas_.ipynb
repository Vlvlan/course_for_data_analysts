```python
df_mcu = pd.concat([filmy,seriale], axis=0)
```

```python
df_mcu
```

```python
df_mcu.drop(columns=["movie_duration", 	"production_budget", 	"opening_weekend", 	"worldwide_box_office", 	"Zarobek na czysto"], inplace=True)
```

```python
df_mcu
```

```python
df_mcu["release_date"] = pd.to_datetime(df_mcu["release_date"], format="%Y-%m-%d")
```

```python
df_mcu
```

```python
df_mcu.info()
```

```python
df_mcu["release_date"] = df_mcu["release_date"].dt.year
```

```python
df_mcu
```

```python
df_mcu["release_data"] = df_mcu['release_data'].astype(int)
```

```python
df_mcu.info()
```

```python
df_covid = df_mcu[df_mcu["release_data"] >= 2019]
```

```python
df_covid.drop(columns=["release_data"], inplace=True)
```

```python
df_covid
```

```python
df_covid.drop(columns=["mcu_phase"], inplace=True)
```

```python
df_covid
```

```python
df_covid["grupa"] = pd.cut(
    df_covid["release_date"],
    bins=[2018, 2023, 2026],
    labels=["grupa_covid", "grupa_po"]
)
```

```python
df_covid
```

```python
df_covid.groupby("grupa").count()
```

```python
df_covid.groupby("grupa").agg(
    średnia_krytyków=("tomato_meter", "mean"),
    średnia_widowni=("audience_score", "mean"),
    Ilość_produkcji=("movie_title", "count")
).round(2)
```
