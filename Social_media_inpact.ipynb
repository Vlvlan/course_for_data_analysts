```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/Social_media_impact_on_life(1).csv")
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.isna().sum()[lambda x: x > 0])
```

```python
df["Student_ID"].duplicated().sum()
```

```python
df.drop(columns=["Student_ID"], inplace=True)
```

```python
df
```

```python
df["Perceived_Stress_Score"] = df['Perceived_Stress_Score'].fillna(df['Perceived_Stress_Score'].mean().round(2))
```

```python
df["Academic_Performance_GPA"] = df['Academic_Performance_GPA'].fillna(df['Academic_Performance_GPA'].mean().round(2))
```

```python
df.describe().round(2).T
```

```python
fig, ax = plt.subplots()
ax.hist(df["Daily_Usage_Hours"])
ax.set_title("Rozdkład dziennego czasu użytkowania")
ax.set_xlabel("Godziny przed ekranem")
ax.set_ylabel("Liczba osob")
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,6))
ax.scatter(df["Age"], df['Daily_Usage_Hours'], alpha=0.6, color="purple")
ax.set_title("Zależnosc między wkiekiem a dziennym czasem korzystania")
ax.set_xlabel("Wiek")
ax.set_ylabel("Dzienny czas spędzony przed ekranem")
ax.grid(True, linestyle="--", alpha=0.5)
plt.show()
```

```python
device_counts = df["Device_Type"].value_counts()
plt.pie(device_counts, autopct="%1.0f%%", radius=0.8, startangle=90)
plt.legend(labels=device_counts.index)
plt.title("Urządzenia wybierane przez osoby 16-26lat")
plt.show()
```

```python
plt.barh(device_counts.index, device_counts.values, color="skyblue")
plt.xlabel("Liczba użyktowników")
plt.ylabel("Rodzaj urządzenia")
plt.show()
```
