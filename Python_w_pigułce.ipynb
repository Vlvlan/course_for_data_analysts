# KALKULATOR

```python
# Stwórz kalkulator który będzie dodawać, odejmować, mnożyć i dzielić
# pamiętaj żeby nie dzielić przez zero
```

```python
liczba_1 = int(input("Podaj liczbe"))
znak = input("Podaj operacje jaką chcesz wykonać (+ - * /) ")
liczba_2 = int(input("Podaj liczbe"))

if znak == "+":
  print(liczba_1 + liczba_2)

elif znak == "-":
  print(liczba_1 - liczba_2)

elif znak == "*":
  print(liczba_1 * liczba_2)

elif znak == "/":
  if liczba_2 == 0:
    print("Nie można dzielić przez zero!!!")
  else:
    print(liczba_1 / liczba_2)
```

```python
def BMI(waga, wzrost):
  wzrost = wzrost ** 2
  return round(waga/wzrost, 2)
```

```python
waga = float(input("Podaj swoją wage    "))
wzrost = float(input("Podaj swój wzrost w metrach   "))
bmi = BMI(waga, wzrost)

if bmi < 18.5:
  print(f"Twoje bmi wynosi {bmi}, masz niedowage")

elif bmi > 18.5 and bmi < 25:
  print(f"Twoje bmi wynosi {bmi}, twoja waga jest prawidłowa")

elif bmi > 25:
  print(f"Twoje bmi wynosi {bmi},  masz nadwage")
```
